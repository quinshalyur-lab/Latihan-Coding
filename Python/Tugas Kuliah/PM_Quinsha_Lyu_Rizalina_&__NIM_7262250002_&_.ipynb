```python
# PM : Komperasi Efesiensi Struktur Kontrol
# Kasir Mini UMKM
# NAMA : Quinsha Lyu Rizalina (NIM : 7262250002)


# PERATURAN TOKO
# Diskon: belanja >= 100.000 -> 10%; >= 50.000 -> 7%; selain itu 0

# CARA 1 : Rantai if-elif
def diskon_ifelif(total):
  if total >= 100000:
    return total * 10 / 100
  elif total >= 50000:
    return total * 7 / 100
  else:
    return 0

def hitung_ongkir(total):
    if total >= 100000:
        return 0
    return 100000

# DAFTAR PRODUK
PRODUK = {"1": ("Cookies Matcha", 8000),
          "2": ("Cookies Coklat", 6000),
          "3": ("Cookies Original", 5000),
          "4": ("Brownies Fudgy Cup", 10000),
          "5": ("Cheese Cake", 15000)}

# PROGRAM KASIR
total = 0
while True:
  kode = input("Kode produk : ")
  if kode not in PRODUK:
    print("Kode tidak ada !")
    continue
  jumlah = int(input("jumlah  : "))
  nama, harga = PRODUK[kode]
  total += harga * jumlah
  if input("Ulangi (y/n): " ).lower() != "y":
    break

diskon = diskon_ifelif(total)
if total >= 100000:
    ongkir = 0
else:
    ongkir = 10000
print("Total Transaksi :", total)
print("Diskon          :", diskon)
print("ongkir          :", ongkir)
print("Jumlah bayar    :", total - diskon + ongkir)

import timeit
import inspect

# CARA 2 : tabel aturan dibaca dengan perulangan
# Urutkan dari batas terbesar dulu
ATURAN_DISKON = [(100000, 10),  (50000, 7)]

def diskon_tabel(total):
  for batas, persen in ATURAN_DISKON:
    if total >= batas:
        return total * persen / 100
  return 0

# 5 data uji: tiap tingkat diskon dan angka tepat di batas
data_uji = [49000, 50000, 99000, 100000, 150000]

print("Belanja | Diskon1 | Diskon2 | Sama? | Ongkir | Bayar")
semua_sama = True
for t in data_uji:
    d1 = diskon_ifelif(t)
    d2 = diskon_tabel(t)
    if t >= 100000:
        ongkir = 0
    else:
        ongkir = 10000
    sama = d1 == d2
    semua_sama = semua_sama and sama
    print (t, "|", d1, "|", d2, "|", sama, "|", ongkir, "|", t - d1 + ongkir)
print("Kedua cara sama:", semua_sama)

# UKUR WAKTU
w1 = timeit.timeit(lambda: [diskon_ifelif(t) for t in data_uji], number=100000)
w2 = timeit.timeit(lambda: [diskon_tabel(t) for t in data_uji], number=100000)
print("waktu cara 1 (if-elif):", round(w1, 4), "detik")
print("waktu cara 2(tabel) :", round(w2, 4), "detik")

# JUMLAH BARIS
baris1 = len(inspect.getsource(diskon_ifelif).splitlines())
baris2 = len(inspect.getsource(diskon_tabel).splitlines())  + 1
print("Baris cara 1:", baris1, "| Baris cara 2:", baris2)
```
